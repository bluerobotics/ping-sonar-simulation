# Receive position geometry

$$ \begin{align*} 
\frac{AB}{v} &= \frac{AP+PB-2R}{c} \\
\\
\text{Let } k=\frac{v}{c}: AB &= kAP+kPB-2kR\\
PB &= \sqrt{r^2+p^2-2rp\cos\theta} = \frac{AB}{k}-AP+2R = \frac{r}{k} - p + 2R\\
r^2+p^2-2rp\cos\theta &= \frac{r^2}{k^2}-\frac{2rp}{k}+\frac{4rR}{k}+p^2-4pR+4R^2 \\
0 &= r^2\left(1-\frac{1}{k^2}\right)+r\left(\frac{2p}{k}-2p\cos\theta-\frac{4R}{k}\right)+4pR-4R^2\\
r &= \frac{-\left(\frac{2p}{k}-2p\cos\theta-\frac{4R}{k}\right)\pm\sqrt{\left(\frac{2p}{k}-2p\cos\theta-\frac{4R}{k}\right)^2-4\left(1-\frac{1}{k^2}\right)\left(4pR-4R^2\right)}}{2\left(1-\frac{1}{k^2}\right)}
\end{align*} $$
