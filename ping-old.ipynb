```python
from ping import Ping
import numpy as np
```

```python
# --- Setup Ping ---
ping_source = Ping(frequency=50e3, 
                   duration=0.5e-3, 
                   amplitude=100, 
                   position=np.array([0,0,0]),
                   sample_rate=5e6)

max_amp = ping_source.original_amplitude
plot_ylim = (-max_amp * 1.1, max_amp * 1.1) # Add 10% padding

# --- Setup Target ---
r_target = 50.0
reflectivity = 0.8      
phase_shift_deg = 180.0 #degrees
```

```python
print(f"Source Level (SL): {ping_source.SL:.2f} dB re 1µPa @ 1m")
ping_source.show(set_title=f"Signal at Source (1m reference)", ylim=plot_ylim)
```

```python
# --- Setup ---
r_target = 50
reflectivity = 0.8      
phase_shift_deg = 180.0 #degrees

print(f"\nStep 2: Propagating {r_target}m to target...")
ping_at_target = ping_source.get_propagated_ping(r_target)
ping_at_target.show(set_title=f"Step 2: Signal at Target (r = {r_target}m)")
```
