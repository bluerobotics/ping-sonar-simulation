```python
import matplotlib.pyplot as plt 
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
from numpy.typing import NDArray
from reflectors.reflector import SeaFloor
```

```python
def plot_spheres_3d(spheres):
    # Unpack the columns (x, y, z, radius)
    x, y, z, r = spheres.T  # (N, 4)

    print(z)

    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Plot points — you’ll at least see sphere *centers*
    scatter = ax.scatter(x, y, z, s=r, alpha=0.6)

    ax.set_xlabel('X (m)')
    ax.set_ylabel('Y (m)')
    ax.set_zlabel('Depth (m)')
    ax.set_title('Seafloor Spheres')

    plt.show()
```

```python
sf = SeaFloor(depth=300, topography='flat', num_spheres=10000)
plot_spheres_3d(sf.seafloor)
sf.seafloor
```
